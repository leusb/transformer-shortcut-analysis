# Shortcuts


Shortcuts: A common definition from Geirhos et al. is essentially a decision rule that works well under the benchmark/training distribution but does not transfer reliably to more challenging conditions. 


So:
- high saliency ≠ shortcut
and
- high label correlation ≠ automatically shortcut.

### Amazon:


For sentiment classification:

- If the task is: infer sentiment from the review text, then "terrible" is exactly the kind of signal you want the model to use.
- If Amazon's binary label was originally derived from product ratings, however: ***one star → negative*** or ***five stars → positive*** is almost a direct encoding of the target. A model can obtain the right answer without really analyzing the rest of the review.


---
Resulting Groups:
1. Robust/ task relevant features: Features that correspond to the intended task. (e.g.: awful
poorly written
very disappointing
highly recommended
waste of money)
2. Shortcut features: Features that are predictive in the observed dataset but provide an easier alternative to the intended reasoning. (e.g. one star
5 stars
1/5)

The model should ideally not become dependent on these


---
***Goal:***

Reduce excessive dependence on shortcut candidates (group 2).
In general: reduce over-realiance on spurious word-label correlations rather than trying to remove predictive word information generally.
